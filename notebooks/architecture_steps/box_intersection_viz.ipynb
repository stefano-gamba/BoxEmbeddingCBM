```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Impostazioni grafiche per una resa pulita da slide
plt.style.use('default')
fig, ax = plt.subplots(figsize=(10, 8))

# Definizioni dei due Box (A e B)
# Box A: X in [2, 6], Y in [1, 4]
z_a = (2, 1)
Z_a = (6, 4)

# Box B: X in [4, 8], Y in [3, 7]
z_b = (4, 3)
Z_b = (8, 7)

# Calcolo dell'intersezione matematica
z_int = (max(z_a[0], z_b[0]), max(z_a[1], z_b[1]))
Z_int = (min(Z_a[0], Z_b[0]), min(Z_a[1], Z_b[1]))

# Disegno il Box A (Blu)
rect_a = patches.Rectangle(z_a, Z_a[0] - z_a[0], Z_a[1] - z_a[1], 
                           linewidth=2, edgecolor='blue', facecolor='blue', alpha=0.2, label='Box A')
ax.add_patch(rect_a)

# Disegno il Box B (Arancione)
rect_b = patches.Rectangle(z_b, Z_b[0] - z_b[0], Z_b[1] - z_b[1], 
                           linewidth=2, edgecolor='orange', facecolor='orange', alpha=0.2, label='Box B')
ax.add_patch(rect_b)

# Disegno l'intersezione (Verde scuro)
rect_int = patches.Rectangle(z_int, Z_int[0] - z_int[0], Z_int[1] - z_int[1], 
                             linewidth=2, edgecolor='green', facecolor='green', alpha=0.6, label='Intersezione A ∩ B')
ax.add_patch(rect_int)

# --- Annotazioni per l'Asse X (Dimensione i=1) ---
# Linee tratteggiate verticali
ax.vlines(x=[z_a[0], z_b[0], Z_a[0], Z_b[0]], ymin=0, ymax=[z_a[1], z_b[1], Z_a[1], Z_b[1]], 
          colors='gray', linestyles='dashed', alpha=0.7)

# Etichette asse X
ax.set_xticks([z_a[0], z_b[0], Z_a[0], Z_b[0]])
ax.set_xticklabels([
    f'$z_{{A,1}}$\n(={z_a[0]})', 
    f'$z_{{B,1}} = \max(z_{{A,1}}, z_{{B,1}})$\n(={z_b[0]})', 
    f'$Z_{{A,1}} = \min(Z_{{A,1}}, Z_{{B,1}})$\n(={Z_a[0]})', 
    f'$Z_{{B,1}}$\n(={Z_b[0]})'
], fontsize=11)

# --- Annotazioni per l'Asse Y (Dimensione i=2) ---
# Linee tratteggiate orizzontali
ax.hlines(y=[z_a[1], z_b[1], Z_a[1], Z_b[1]], xmin=0, xmax=[z_a[0], z_b[0], Z_a[0], Z_b[0]], 
          colors='gray', linestyles='dashed', alpha=0.7)

# Etichette asse Y
ax.set_yticks([z_a[1], z_b[1], Z_a[1], Z_b[1]])
ax.set_yticklabels([
    f'$z_{{A,2}}$ (={z_a[1]})', 
    f'$z_{{B,2}} = \max(z_{{A,2}}, z_{{B,2}})$ (={z_b[1]})', 
    f'$Z_{{A,2}} = \min(Z_{{A,2}}, Z_{{B,2}})$ (={Z_a[1]})', 
    f'$Z_{{B,2}}$ (={Z_b[1]})'
], fontsize=11)

# Configurazione finale del plot
ax.set_xlim(0, 9)
ax.set_ylim(0, 8)
ax.set_xlabel('Dimensione 1', fontsize=14, fontweight='bold')
ax.set_ylabel('Dimensione 2', fontsize=14, fontweight='bold')
ax.set_title('Rappresentazione Geometrica: Intersezione e Coordinate', fontsize=16, fontweight='bold', pad=20)
ax.legend(loc='upper left', fontsize=12)
ax.grid(True, linestyle=':', alpha=0.5)

# Mostra il grafico
plt.tight_layout()
plt.show()
```
