```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Impostazioni grafiche per una resa pulita da slide
plt.style.use('default')
fig, ax = plt.subplots(figsize=(10, 8))

# Coordinate dipendenti dai parametri latenti theta
# Dimensione 1 (Asse X)
z_1 = 2
Z_1 = 6
# Dimensione 2 (Asse Y)
z_2 = 1
Z_2 = 4

# 1. Evidenzio gli intervalli mono-dimensionali direttamente sugli assi
# Intervallo asse X (Blu) - lo disegno esattamente su y=0
ax.plot([z_1, Z_1], [0, 0], color='blue', linewidth=8, solid_capstyle='butt', label='Intervallo Dim 1: $[z_1, Z_1]$')
# Intervallo asse Y (Rosso) - lo disegno esattamente su x=0
ax.plot([0, 0], [z_2, Z_2], color='red', linewidth=8, solid_capstyle='butt', label='Intervallo Dim 2: $[z_2, Z_2]$')

# 2. Linee di proiezione tratteggiate
# Mostrano come l'intervallo X si estende nello spazio Y
ax.vlines(x=[z_1, Z_1], ymin=0, ymax=Z_2, colors='blue', linestyles='dashed', alpha=0.5)
# Mostrano come l'intervallo Y si estende nello spazio X
ax.hlines(y=[z_2, Z_2], xmin=0, xmax=Z_1, colors='red', linestyles='dashed', alpha=0.5)

# 3. Disegno il Box (Il risultato del Prodotto Cartesiano)
box = patches.Rectangle((z_1, z_2), Z_1 - z_1, Z_2 - z_2, 
                        linewidth=2, edgecolor='purple', facecolor='purple', alpha=0.3)
ax.add_patch(box)

# 4. Formattazione degli Assi (stile piano cartesiano puro)
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)

# 5. Etichette personalizzate che richiamano la dipendenza da theta
ax.set_xticks([z_1, Z_1])
ax.set_xticklabels([f'$z_1(\\theta)$\n(= {z_1})', f'$Z_1(\\theta)$\n(= {Z_1})'], fontsize=12)

ax.set_yticks([z_2, Z_2])
ax.set_yticklabels([f'$z_2(\\theta)$\n(= {z_2})', f'$Z_2(\\theta)$\n(= {Z_2})'], fontsize=12)

ax.set_xlim(-1, 8)
ax.set_ylim(-1, 6)

# Testo esplicativo centrale
ax.text((z_1 + Z_1)/2, (z_2 + Z_2)/2, 
        '$B(\\theta) = [z_1, Z_1] \\times [z_2, Z_2]$', 
        fontsize=16, fontweight='bold', ha='center', va='center', color='black')

# Titoli e Legenda
ax.set_xlabel('Dimensione 1', fontsize=14, fontweight='bold', color='blue', loc='right')
ax.set_ylabel('Dimensione 2', fontsize=14, fontweight='bold', color='red', loc='top')
ax.set_title("Definizione del Box: Prodotto Cartesiano di Intervalli", fontsize=16, fontweight='bold', pad=20)
ax.legend(loc='upper right', fontsize=12)
ax.grid(True, linestyle=':', alpha=0.3)

plt.tight_layout()
plt.show()
```
